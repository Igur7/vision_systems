  Wykorzystując funkcję `mesh` zwizualizuj filtr.
  Sprawdź jak parametr `odchylenie standardowe` wpływa na `kształt` filtru.

  Uwaga. W OpenCV dostępna jest *dedykowana* funkcja do filtracji Gaussa - `GaussianBlur`.
  Proszę na jednym przykładzie porównać jej działanie z użytym wyżej rozwiązaniem.

10. Wykonaj filtrację dla wybranych (2--3) wartości odchylenia standardowego.


```python
def fgaussian(size, sigma):
     m = n = size
     h, k = m//2, n//2
     x, y = np.mgrid[-h:h+1, -k:k+1]
     g = np.exp(-(x**2 + y**2)/(2*sigma**2))
     return g /g.sum() 
    
    
def mesh(fun, size):
    fig = plt.figure()
    ax = fig.add_subplot(projection='3d')
    

    X = np.arange(-size//2, size//2, 1)
    Y = np.arange(-size//2, size//2, 1)
    X, Y = np.meshgrid(X, Y)
    Z = fun
    
    ax.plot_surface(X, Y, Z)
    
    plt.show()
    
"""8. Skuteczną i często wykorzystywaną maską jest tzw. maska Gasussa.
Jest to zbiór liczb, które aproksymują dwuwymiarowy rozkład Gaussa. 
Parametrem jest odchylenie standardowe i rozmiar maski.

9. Wykorzystując przygotowaną funkcję `fgaussian` stwórz maskę o rozmiarze $5 \times 5$ i odchyleniu standardowym 0.5.
  Wykorzystując funkcję `mesh` zwizualizuj filtr.
  Sprawdź jak parametr `odchylenie standardowe` wpływa na `kształt` filtru.

  Uwaga. W OpenCV dostępna jest *dedykowana* funkcja do filtracji Gaussa - `GaussianBlur`.
  Proszę na jednym przykładzie porównać jej działanie z użytym wyżej rozwiązaniem.

10. Wykonaj filtrację dla wybranych (2--3) wartości odchylenia standardowego."""


gaussian_mask = fgaussian(5, 0.5)
mesh(gaussian_mask, 5)
gaussian_mask = fgaussian(5, 1)
mesh(gaussian_mask, 5)
gaussian_mask = fgaussian(5, 2)
mesh(gaussian_mask, 5)
cv2_gaussian = cv2.GaussianBlur(lena, (5, 5), 5)
gaussian_lena = cv2.filter2D(lena, -1, fgaussian(5, 5))
plt.figure(figsize=(15,5))  
plt.subplot(1,3,1)
plt.imshow(lena, cmap='gray')
plt.title('Original')
plt.axis('off')
plt.subplot(1,3,2)
plt.imshow(cv2_gaussian, cmap='gray')
plt.title('GaussianBlur (OpenCV)')
plt.axis('off')
plt.subplot(1,3,3)
plt.imshow(gaussian_lena, cmap='gray')
plt.title('Gaussian Filter (Custom)')
plt.axis('off')
plt.show()
```

Podstawowa różnica pomiędzy filtrami liniowymi, a nieliniowymi polega na tym, że przy filtracji liniowej na nową wartość piksela ma wpływ wartość wszystkich pikseli z otoczenia (np. uśrednianie, czasem ważone), natomiast w przypadku filtracji nieliniowej jako nowy piksel wybierana jest któraś z wartości otoczenia - według jakiegoś wskaźnika (wartość największa, najmniejsza czy właśnie mediana).

4. Wykonaj konwolucję obrazu z maską (`c2.filter2D`). Przed wyświetleniem, wynikowy obraz należy poddać normalizacji (występują ujemne wartości). Najczęściej wykonuje się jedną z dwóch operacji:

\end{equation}

Przeprowadź detekcję krawędzi za pomocą gradientu Sobela. 

```python
R1 = np.array([[0, 0, 0],[-1,0,0],[0,1,0]])
R2 = np.array([[0, 0, 0],[0,0,-1],[0,1,0]])
P1 = np.array([[-1,0,1],[-1,0,1],[-1,0,1]])
P2 = np.array([[-1,-1,-1],[0,0,0],[1,1,1]])
S1 = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
S2 = np.array([[-1,-2,-1],[0,0,0],[1,2,1]])

def apply_filter(image, kernel):
    plt.figure(figsize=(15,5))
    plt.subplot(1,3,1)
    plt.imshow(image, cmap='gray')
    plt.title('Original')
    plt.axis('off')
    filtered_image = cv2.filter2D(image.astype(np.float64), -1, kernel)
    filtered_image_add128 = np.clip(filtered_image + 128, 0, 255)
    plt.subplot(1,3,2)
    plt.imshow(filtered_image_add128, cmap='gray')
    plt.title('Filtered + 128')
    plt.axis('off')
    filtered_image_abs = np.abs(filtered_image)
    plt.subplot(1,3,3)
    plt.imshow(filtered_image_abs, cmap='gray')
    plt.title('Absolute difference')
    plt.axis('off')
    plt.tight_layout()
    plt.show()
    
apply_filter(kw, R1)
apply_filter(kw, R2)
apply_filter(kw, P1)
apply_filter(kw, P2)
apply_filter(kw, S1)
apply_filter(kw, S2)


```

5. Na podstawie dwóch ortogonalnych masek np. Sobela można stworzyć tzw. filtr kombinowany - pierwiastek kwadratowy z sumy kwadratów gradientów:
\begin{equation}
OW = \sqrt{(O * S1)^2 + (O * S2)^2}
\tag{5}
\end{equation}
gdzie:  $OW$ - obraz wyjściowy, $O$ - obraz oryginalny (wejściowy), $S1,S2$ - maski Sobela, $*$ - operacja konwolucji.

Zaimplementuj filtr kombinowany.

Uwaga. Proszę zwrócić uwagę na konieczność zmiany formatu danych obrazu wejściowego - na typ znakiem



```python
def creaate_cpmbined(image_original,S1,S2):
    return np.sqrt(cv2.filter2D(image_original.astype(np.float64), -1, S1)**2 + cv2.filter2D(image_original.astype(np.float64), -1, S2)**2)

plt.figure(figsize=(15,5))
plt.subplot(1,3,1)
plt.imshow(kw, cmap='gray')
plt.title('Original')
plt.axis('off')
combined_kw = creaate_cpmbined(kw, S1, S2)
plt.subplot(1,3,2)
plt.imshow(combined_kw, cmap='gray')
plt.title('Combined Filtered')
plt.axis('off')
plt.subplot(1,3,3)
plt.imshow(np.abs(kw - combined_kw), cmap='gray')
plt.title('Difference')
plt.axis('off')
plt.show()
```

6. Istnieje alternatywna wersja filtra kombinowanego, która zamiast pierwiastka z sumy kwadratów wykorzystuje sumę modułów (prostsze obliczenia). 
Zaimplementuj tę wersję. 

```python
def alternative_combined(image_original, S1, S2):
    filtered_S1 = cv2.filter2D(image_original.astype(np.float64), -1, S1)
    filtered_S2 = cv2.filter2D(image_original.astype(np.float64), -1, S2)
    return np.abs(filtered_S1) + np.abs(filtered_S2)

plt.figure(figsize=(15,5))
plt.subplot(1,3,1)
plt.imshow(kw, cmap='gray')
plt.title('Original')
plt.axis('off')
combined_kw = alternative_combined(kw, S1, S2)
plt.subplot(1,3,2)
plt.imshow(combined_kw, cmap='gray')   
plt.title('Alternative Combined Filtered')
plt.axis('off')
plt.subplot(1,3,3)
plt.imshow(np.abs(kw - combined_kw), cmap='gray')
plt.title('Difference')
plt.axis('off')
plt.show()
```
